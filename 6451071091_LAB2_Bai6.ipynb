```python
# BÀI 6 - CÂU a
# UPLOAD VÀ ĐỌC DỮ LIỆU MUSHROOM

import pandas as pd
import numpy as np
from google.colab import files

print("Họ tên: Phạm Hải Yến")
print("MSSV: 6451071091")
print("Bài 6 - Câu a")
print("-" * 50)

# 1. Upload file
uploaded = files.upload()
file_name = list(uploaded.keys())[0]

print("File đã upload:", file_name)

# 2. Đọc dữ liệu
mushroom_df = pd.read_csv(
    file_name,
    sep=",",
    header="infer"
)

# 3. Chuẩn hóa tên cột
mushroom_df.columns = (
    mushroom_df.columns
    .astype(str)
    .str.strip()
)

print("\nKích thước dữ liệu:", mushroom_df.shape)

print("\nTên các thuộc tính:")
print(mushroom_df.columns.tolist())

print("\n5 dòng đầu tiên:")
display(mushroom_df.head())
```

```python
# BÀI 6 - CÂU 4
# KHẢO SÁT DỮ LIỆU MUSHROOM VÀ TƯƠNG QUAN

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Họ tên: Phạm Hải Yến")
print("MSSV: 6451071091")
print("Bài 6 - Câu 4")
print("-" * 50)

# Kiểm tra dữ liệu
print("Kích thước dữ liệu:", mushroom_df.shape)

print("\nSố lượng dữ liệu thiếu:")
print(mushroom_df.isnull().sum())

print("\nSố lượng mẫu theo class:")
print(mushroom_df["class"].value_counts())

# Mã hóa dữ liệu phân loại để khảo sát tương quan
mushroom_encoded = mushroom_df.copy()

for col in mushroom_encoded.columns:
    mushroom_encoded[col] = (
        mushroom_encoded[col]
        .astype("category")
        .cat.codes
    )

# Tính ma trận tương quan
corr_matrix = mushroom_encoded.corr(method="pearson")

print("\nMa trận tương quan:")
display(corr_matrix)

# Vẽ heatmap
plt.figure(figsize=(16, 12))

sns.heatmap(
    corr_matrix,
    cmap="coolwarm",
    annot=False
)

plt.title("Ma trận tương quan - Mushroom")
plt.show()
```

```python
# BÀI 6 - CÂU 5
# TÁCH X VÀ y

print("Họ tên: Phạm Hải Yến")
print("MSSV: 6451071091")
print("Bài 6 - Câu 5")
print("-" * 50)

# X: các thuộc tính đầu vào
X = mushroom_df.drop(columns=["class"])

# y: thuộc tính quyết định
y = mushroom_df["class"]

print("Kích thước X:", X.shape)
print("Kích thước y:", y.shape)

print("\nCác thuộc tính đầu vào:")
print(X.columns.tolist())

print("\nCác lớp class:")
print(y.value_counts())

print("\nÝ nghĩa class:")
print("p: Nấm độc (poisonous)")
print("e: Nấm ăn được (edible)")
```

```python
# BÀI 6 - CÂU 6
# CHIA DỮ LIỆU TRAIN VÀ TEST

from sklearn.model_selection import train_test_split

print("Họ tên: Phạm Hải Yến")
print("MSSV: 6451071091")
print("Bài 6 - Câu 6")
print("-" * 50)

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=42
)

print("Kích thước X_train:", X_train.shape)
print("Kích thước X_test:", X_test.shape)

print("Kích thước y_train:", y_train.shape)
print("Kích thước y_test:", y_test.shape)

print("\nTỷ lệ train:",
      len(X_train) / len(X))

print("Tỷ lệ test:",
      len(X_test) / len(X))
```

```python
# BÀI 6 - CÂU 7
# XÂY DỰNG CÂY QUYẾT ĐỊNH ID3

import pandas as pd
import matplotlib.pyplot as plt

from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report
)

print("Họ tên: Phạm Hải Yến")
print("MSSV: 6451071091")
print("Bài 6 - Câu 7 - ID3")
print("-" * 50)

# Mã hóa thuộc tính phân loại
X_train_encoded = pd.get_dummies(
    X_train,
    dtype=int
)

X_test_encoded = pd.get_dummies(
    X_test,
    dtype=int
)

# Đồng bộ cột train và test
X_train_encoded, X_test_encoded = X_train_encoded.align(
    X_test_encoded,
    join="outer",
    axis=1,
    fill_value=0
)

# Xây dựng mô hình ID3 theo tiêu chí Entropy
id3_model = DecisionTreeClassifier(
    criterion="entropy",
    random_state=42
)

# Huấn luyện
id3_model.fit(X_train_encoded, y_train)

# Dự đoán
y_pred_id3 = id3_model.predict(X_test_encoded)

# Đánh giá
accuracy_id3 = accuracy_score(
    y_test,
    y_pred_id3
)

print("Độ chính xác ID3:", accuracy_id3)

print("\nMa trận nhầm lẫn:")
print(confusion_matrix(y_test, y_pred_id3))

print("\nBáo cáo phân loại:")
print(classification_report(
    y_test,
    y_pred_id3,
    zero_division=0
))

# Vẽ cây quyết định
plt.figure(figsize=(20, 10))

plot_tree(
    id3_model,
    feature_names=X_train_encoded.columns,
    class_names=[str(c) for c in id3_model.classes_],
    filled=True,
    rounded=True,
    max_depth=3,
    fontsize=8
)

plt.title("Cây quyết định ID3 - Mushroom")
plt.show()
```

```python
# BÀI 6 - CÂU 8
# XÂY DỰNG CÂY QUYẾT ĐỊNH CART

import matplotlib.pyplot as plt

from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report
)

print("Họ tên: Phạm Hải Yến")
print("MSSV: 6451071091")
print("Bài 6 - Câu 8 - CART")
print("-" * 50)

# Khởi tạo mô hình CART
cart_model = DecisionTreeClassifier(
    criterion="gini",
    random_state=42
)

# Huấn luyện
cart_model.fit(X_train_encoded, y_train)

# Dự đoán
y_pred_cart = cart_model.predict(X_test_encoded)

# Đánh giá
accuracy_cart = accuracy_score(
    y_test,
    y_pred_cart
)

print("Độ chính xác CART:", accuracy_cart)

print("\nMa trận nhầm lẫn:")
print(confusion_matrix(y_test, y_pred_cart))

print("\nBáo cáo phân loại:")
print(classification_report(
    y_test,
    y_pred_cart,
    zero_division=0
))

# Vẽ cây quyết định
plt.figure(figsize=(20, 10))

plot_tree(
    cart_model,
    feature_names=X_train_encoded.columns,
    class_names=[str(c) for c in cart_model.classes_],
    filled=True,
    rounded=True,
    max_depth=3,
    fontsize=8
)

plt.title("Cây quyết định CART - Mushroom")
plt.show()
```

```python
# BÀI 6 - CÂU 9
# XÂY DỰNG MÔ HÌNH NAIVE BAYES

import matplotlib.pyplot as plt

from sklearn.naive_bayes import BernoulliNB
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report,
    ConfusionMatrixDisplay
)

print("Họ tên: Phạm Hải Yến")
print("MSSV: 6451071091")
print("Bài 6 - Câu 9 - Naive Bayes")
print("-" * 50)

# Khởi tạo mô hình Bernoulli Naive Bayes
nb_model = BernoulliNB()

# Huấn luyện
nb_model.fit(X_train_encoded, y_train)

# Dự đoán
y_pred_nb = nb_model.predict(X_test_encoded)

# Đánh giá
accuracy_nb = accuracy_score(
    y_test,
    y_pred_nb
)

print("Độ chính xác Naive Bayes:", accuracy_nb)

print("\nMa trận nhầm lẫn:")
cm_nb = confusion_matrix(y_test, y_pred_nb)
print(cm_nb)

print("\nBáo cáo phân loại:")
print(classification_report(
    y_test,
    y_pred_nb,
    zero_division=0
))

# Vẽ ma trận nhầm lẫn
disp = ConfusionMatrixDisplay(
    confusion_matrix=cm_nb,
    display_labels=nb_model.classes_
)

disp.plot()
plt.title("Ma trận nhầm lẫn - Naive Bayes")
plt.show()
```

```python
# BÀI 6 - CÂU 10
# SO SÁNH ID3, CART VÀ NAIVE BAYES

import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

print("Họ tên: Phạm Hải Yến")
print("MSSV: 6451071091")
print("Bài 6 - Câu 10")
print("-" * 50)

# Các mô hình và kết quả dự đoán
models = {
    "ID3": y_pred_id3,
    "CART": y_pred_cart,
    "Naive Bayes": y_pred_nb
}

# Tính các chỉ số đánh giá
results = []

for model_name, y_pred in models.items():
    results.append({
        "Mô hình": model_name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(
            y_test,
            y_pred,
            average="weighted",
            zero_division=0
        ),
        "Recall": recall_score(
            y_test,
            y_pred,
            average="weighted",
            zero_division=0
        ),
        "F1-score": f1_score(
            y_test,
            y_pred,
            average="weighted",
            zero_division=0
        )
    })

comparison = pd.DataFrame(results)

print("Bảng so sánh kết quả:")
display(comparison)

# Vẽ biểu đồ so sánh Accuracy
plt.figure(figsize=(8, 5))

plt.bar(
    comparison["Mô hình"],
    comparison["Accuracy"]
)

plt.title("So sánh Accuracy - Mushroom")
plt.xlabel("Mô hình")
plt.ylabel("Accuracy")
plt.ylim(0, 1)

for i, value in enumerate(comparison["Accuracy"]):
    plt.text(
        i,
        value + 0.01,
        f"{value:.4f}",
        ha="center"
    )

plt.show()

# Tìm mô hình có Accuracy cao nhất
best_model = comparison.loc[
    comparison["Accuracy"].idxmax()
]

print("\nMô hình có Accuracy cao nhất:",
      best_model["Mô hình"])

print("Accuracy:",
      best_model["Accuracy"])
```

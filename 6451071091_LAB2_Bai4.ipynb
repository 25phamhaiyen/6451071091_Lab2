```python
# BÀI 4 - CÂU 3
# UPLOAD, ĐỌC VÀ KHẢO SÁT DỮ LIỆU RED WINE QUALITY

import pandas as pd
import numpy as np
from google.colab import files

print("Họ tên: Phạm Hải Yến")
print("MSSV: 6451071091")
print("Bài 4 - Câu 3")
print("-" * 50)

# 1. Upload file dữ liệu
uploaded = files.upload()

file_name = list(uploaded.keys())[0]

print("\nFile đã upload:", file_name)

# 2. Đọc dữ liệu
# File hiện tại sử dụng dấu phẩy để phân cách
wine_df = pd.read_csv(
    file_name,
    sep=","
)

# 3. Khảo sát dữ liệu
print("\nKích thước dữ liệu:", wine_df.shape)

print("\nTên các thuộc tính:")
print(wine_df.columns.tolist())

print("\nThông tin dữ liệu:")
wine_df.info()

print("\n5 dòng dữ liệu đầu tiên:")
display(wine_df.head())

print("\nThống kê mô tả:")
display(wine_df.describe())

print("\nSố lượng dữ liệu thiếu:")
print(wine_df.isnull().sum())
```

```python
# BÀI 4 - CÂU 4
# TIỀN XỬ LÝ VÀ KHẢO SÁT TƯƠNG QUAN

import matplotlib.pyplot as plt
import seaborn as sns

print("Họ tên: Phạm Hải Yến")
print("MSSV: 6451071091")
print("Bài 4 - Câu 4")
print("-" * 50)

# Xóa dòng có dữ liệu thiếu
wine_df = wine_df.dropna().copy()

# Kiểm tra dữ liệu trùng lặp
print("Số dòng trùng lặp:",
      wine_df.duplicated().sum())

# Tính tương quan Pearson
corr_matrix = wine_df.corr(method="pearson")

print("\nMa trận tương quan:")
display(corr_matrix)

# Vẽ heatmap
plt.figure(figsize=(12, 8))

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Ma trận tương quan Pearson - Red Wine Quality")
plt.show()

# Tìm các cặp thuộc tính có tương quan cao
threshold = 0.8

high_corr = []

for i in range(len(corr_matrix.columns)):
    for j in range(i + 1, len(corr_matrix.columns)):
        value = corr_matrix.iloc[i, j]

        if abs(value) >= threshold:
            high_corr.append((
                corr_matrix.columns[i],
                corr_matrix.columns[j],
                value
            ))

print("\nCác cặp thuộc tính có tương quan cao:")
for col1, col2, value in high_corr:
    print(f"{col1} - {col2}: {value:.3f}")
```

```python
# BÀI 4 - CÂU 5
# TÁCH X VÀ y

print("Họ tên: Phạm Hải Yến")
print("MSSV: 6451071091")
print("Bài 4 - Câu 5")
print("-" * 50)

# Thuộc tính quyết định
target_column = "quality"

# X: các thuộc tính đầu vào
X = wine_df.drop(columns=[target_column])

# y: nhãn cần dự đoán
y = wine_df[target_column]

print("Kích thước X:", X.shape)
print("Kích thước y:", y.shape)

print("\nCác thuộc tính đầu vào:")
print(X.columns.tolist())

print("\nCác giá trị quality:")
print(sorted(y.unique()))

print("\nSố lượng mẫu theo từng quality:")
print(y.value_counts().sort_index())
```

```python
# BÀI 4 - CÂU 6
# CHIA DỮ LIỆU TRAIN VÀ TEST

from sklearn.model_selection import train_test_split

print("Họ tên: Phạm Hải Yến")
print("MSSV: 6451071091")
print("Bài 4 - Câu 6")
print("-" * 50)

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=42,
    stratify=y
)

print("Kích thước X_train:", X_train.shape)
print("Kích thước X_test:", X_test.shape)

print("Kích thước y_train:", y_train.shape)
print("Kích thước y_test:", y_test.shape)

print("\nTỷ lệ train:",
      len(X_train) / len(X))

print("Tỷ lệ test:",
      len(X_test) / len(X))
```

```python
# BÀI 4 - CÂU 7
# XÂY DỰNG CÂY QUYẾT ĐỊNH ID3

import matplotlib.pyplot as plt

from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report
)

print("Họ tên: Phạm Hải Yến")
print("MSSV: 6451071091")
print("Bài 4 - Câu 7 - ID3")
print("-" * 50)

# Khởi tạo mô hình ID3
id3_model = DecisionTreeClassifier(
    criterion="entropy",
    random_state=42
)

# Huấn luyện
id3_model.fit(X_train, y_train)

# Dự đoán
y_pred_id3 = id3_model.predict(X_test)

# Độ chính xác
accuracy_id3 = accuracy_score(
    y_test,
    y_pred_id3
)

print("Độ chính xác ID3:", accuracy_id3)

print("\nMa trận nhầm lẫn:")
print(confusion_matrix(y_test, y_pred_id3))

print("\nBáo cáo phân loại:")
print(classification_report(
    y_test,
    y_pred_id3,
    zero_division=0
))

# Vẽ cây quyết định
plt.figure(figsize=(20, 10))

plot_tree(
    id3_model,
    feature_names=X.columns,
    class_names=[str(c) for c in id3_model.classes_],
    filled=True,
    rounded=True,
    max_depth=3,
    fontsize=8
)

plt.title("Cây quyết định ID3 - Red Wine Quality")
plt.show()
```

```python
# BÀI 4 - CÂU 8
# XÂY DỰNG CÂY QUYẾT ĐỊNH CART

from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report
)

print("Họ tên: Phạm Hải Yến")
print("MSSV: 6451071091")
print("Bài 4 - Câu 8 - CART")
print("-" * 50)

# Khởi tạo mô hình CART
cart_model = DecisionTreeClassifier(
    criterion="gini",
    random_state=42
)

# Huấn luyện
cart_model.fit(X_train, y_train)

# Dự đoán
y_pred_cart = cart_model.predict(X_test)

# Đánh giá
accuracy_cart = accuracy_score(
    y_test,
    y_pred_cart
)

print("Độ chính xác CART:", accuracy_cart)

print("\nMa trận nhầm lẫn:")
print(confusion_matrix(y_test, y_pred_cart))

print("\nBáo cáo phân loại:")
print(classification_report(
    y_test,
    y_pred_cart,
    zero_division=0
))

# Vẽ cây CART
plt.figure(figsize=(20, 10))

plot_tree(
    cart_model,
    feature_names=X.columns,
    class_names=[str(c) for c in cart_model.classes_],
    filled=True,
    rounded=True,
    max_depth=3,
    fontsize=8
)

plt.title("Cây quyết định CART - Red Wine Quality")
plt.show()
```

```python
# BÀI 4 - CÂU 9
# XÂY DỰNG MÔ HÌNH NAIVE BAYES

from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report,
    ConfusionMatrixDisplay
)

import matplotlib.pyplot as plt

print("Họ tên: Phạm Hải Yến")
print("MSSV: 6451071091")
print("Bài 4 - Câu 9 - Naive Bayes")
print("-" * 50)

# Khởi tạo mô hình
nb_model = GaussianNB()

# Huấn luyện
nb_model.fit(X_train, y_train)

# Dự đoán
y_pred_nb = nb_model.predict(X_test)

# Đánh giá
accuracy_nb = accuracy_score(
    y_test,
    y_pred_nb
)

print("Độ chính xác Naive Bayes:", accuracy_nb)

print("\nMa trận nhầm lẫn:")
cm_nb = confusion_matrix(y_test, y_pred_nb)
print(cm_nb)

print("\nBáo cáo phân loại:")
print(classification_report(
    y_test,
    y_pred_nb,
    zero_division=0
))

# Vẽ ma trận nhầm lẫn
disp = ConfusionMatrixDisplay(
    confusion_matrix=cm_nb,
    display_labels=nb_model.classes_
)

disp.plot()
plt.title("Ma trận nhầm lẫn - Naive Bayes")
plt.show()
```

```python
# BÀI 4 - CÂU 10
# SO SÁNH CÁC MÔ HÌNH

import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

print("Họ tên: Phạm Hải Yến")
print("MSSV: 6451071091")
print("Bài 4 - Câu 10")
print("-" * 50)

# Danh sách mô hình và kết quả dự đoán
models = {
    "ID3": y_pred_id3,
    "CART": y_pred_cart,
    "Naive Bayes": y_pred_nb
}

# Tính các chỉ số đánh giá
results = []

for model_name, y_pred in models.items():
    results.append({
        "Mô hình": model_name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(
            y_test,
            y_pred,
            average="weighted",
            zero_division=0
        ),
        "Recall": recall_score(
            y_test,
            y_pred,
            average="weighted",
            zero_division=0
        ),
        "F1-score": f1_score(
            y_test,
            y_pred,
            average="weighted",
            zero_division=0
        )
    })

comparison = pd.DataFrame(results)

print("Bảng so sánh kết quả:")
display(comparison)

# Vẽ biểu đồ so sánh Accuracy
plt.figure(figsize=(8, 5))

plt.bar(
    comparison["Mô hình"],
    comparison["Accuracy"]
)

plt.title("So sánh Accuracy các mô hình")
plt.xlabel("Mô hình")
plt.ylabel("Accuracy")
plt.ylim(0, 1)

for i, value in enumerate(comparison["Accuracy"]):
    plt.text(
        i,
        value + 0.01,
        f"{value:.4f}",
        ha="center"
    )

plt.show()

# Xác định mô hình tốt nhất theo Accuracy
best_model = comparison.loc[
    comparison["Accuracy"].idxmax()
]

print("\nMô hình có Accuracy cao nhất:",
      best_model["Mô hình"])

print("Accuracy:",
      best_model["Accuracy"])
```

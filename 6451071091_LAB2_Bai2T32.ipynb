```python
# BÀI 1 - DỰ ĐOÁN DOANH SỐ ĐỒ CHƠI
# Họ tên: Phạm Hải Yến
# MSSV: 6451071091

import pandas as pd
import numpy as np

from sklearn.preprocessing import OrdinalEncoder
from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import CategoricalNB
from sklearn.metrics import (
    confusion_matrix,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

# 1. Nhập dữ liệu huấn luyện theo bảng trang 24
data1 = {
    "Loai": [
        "Điều khiển", "Xếp hình", "Xếp hình",
        "Điều khiển", "Búp bê", "Điều khiển",
        "Búp bê", "Điều khiển", "Xếp hình",
        "Xếp hình", "Búp bê", "Xếp hình",
        "Điều khiển", "Búp bê", "Búp bê"
    ],
    "So_mau": [3, 5, 7, 5, 3, 5, 5, 7, 7, 3, 3, 3, 5, 5, 7],
    "Kich_thuoc": [
        "Nhỏ", "Vừa", "To", "Nhỏ", "Vừa",
        "Vừa", "To", "Vừa", "To", "To",
        "Nhỏ", "Nhỏ", "To", "Vừa", "To"
    ],
    "Chat_lieu": [
        "Nhựa PP", "Cao su", "Nhựa PP", "Cao su", "Nhựa PP",
        "Nhựa PP", "Nhựa PP", "Cao su", "Cao su", "Nhựa PP",
        "Cao su", "Nhựa PP", "Nhựa PP", "Nhựa PP", "Nhựa PP"
    ],
    "Doanh_so": [
        "Cao", "Thấp", "Thấp", "Thấp", "Thấp",
        "Cao", "Cao", "Thấp", "Cao", "Thấp",
        "Thấp", "Cao", "Thấp", "Cao", "Cao"
    ]
}

df1 = pd.DataFrame(data1)

print("Họ tên: Phạm Hải Yến")
print("MSSV: 6451071091")
print("\nDữ liệu huấn luyện:")
print(df1)

# 2. Tách thuộc tính và nhãn
X1 = df1.drop("Doanh_so", axis=1)
y1 = df1["Doanh_so"]

# 3. Mã hóa thuộc tính dạng chữ
encoder1 = OrdinalEncoder()
X1_encoded = encoder1.fit_transform(X1)

# 4. Khởi tạo và huấn luyện mô hình
id3_1 = DecisionTreeClassifier(
    criterion="entropy",
    random_state=42
)

cart_1 = DecisionTreeClassifier(
    criterion="gini",
    random_state=42
)

nb_1 = CategoricalNB()

id3_1.fit(X1_encoded, y1)
cart_1.fit(X1_encoded, y1)
nb_1.fit(X1_encoded, y1)

# 5. Nhập các mẫu cần dự đoán theo đề
test1 = pd.DataFrame({
    "Loai": ["Búp bê", "Xếp hình", "Điều khiển"],
    "So_mau": [3, 5, 3],
    "Kich_thuoc": ["To", "To", "Vừa"],
    "Chat_lieu": ["Cao su", "Nhựa PP", "Cao su"]
})

# Doanh số thực tế theo yêu cầu e
y1_test = ["Thấp", "Thấp", "Cao"]

X1_test = encoder1.transform(test1)

# 6. Hàm đánh giá mô hình
def danh_gia(y_true, y_pred, ten_mo_hinh):
    print("\n==========", ten_mo_hinh, "==========")

    print("Nhãn thực tế:", y_true)
    print("Nhãn dự đoán:", y_pred)

    print("\nConfusion Matrix:")
    print(confusion_matrix(
        y_true,
        y_pred,
        labels=["Thấp", "Cao"]
    ))

    precision = precision_score(
        y_true, y_pred,
        average="weighted",
        zero_division=0
    )

    recall = recall_score(
        y_true, y_pred,
        average="weighted",
        zero_division=0
    )

    f_measure = f1_score(
        y_true, y_pred,
        average="weighted",
        zero_division=0
    )

    print("Precision:", round(precision, 4))
    print("Recall:", round(recall, 4))
    print("F-measure:", round(f_measure, 4))

    print("\nClassification Report:")
    print(classification_report(
        y_true,
        y_pred,
        zero_division=0
    ))

# 7. Dự đoán và tính F-measure
pred_id3_1 = id3_1.predict(X1_test)
pred_cart_1 = cart_1.predict(X1_test)
pred_nb_1 = nb_1.predict(X1_test)

danh_gia(y1_test, pred_id3_1, "ID3")
danh_gia(y1_test, pred_cart_1, "CART")
danh_gia(y1_test, pred_nb_1, "Naive Bayes")
```

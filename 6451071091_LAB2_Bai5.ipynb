```python
# BÀI 5 - CÂU a
# ĐỌC VÀ TIỀN XỬ LÝ DỮ LIỆU HEART DISEASE

import pandas as pd
import numpy as np
from google.colab import files

print("Họ tên: Phạm Hải Yến")
print("MSSV: 6451071091")
print("Bài 5 - Câu a")
print("-" * 50)

# 1. Upload file dữ liệu
uploaded = files.upload()
file_name = list(uploaded.keys())[0]

print("File đã upload:", file_name)

# 2. Đọc dữ liệu
heart_df = pd.read_csv(
    file_name,
    sep=None,
    engine="python"
)

# 3. Danh sách tên cột theo đề
column_names = [
    "age",
    "sex",
    "chest_pain_type",
    "resting_blood_pressure",
    "cholesterol",
    "fasting_blood_sugar",
    "rest_ecg",
    "max_heart_rate_achieved",
    "exercise_induced_angina",
    "st_depression",
    "st_slope",
    "num_major_vessels",
    "thalassemia",
    "target"
]

# 4. Chuẩn hóa tên cột
heart_df.columns = heart_df.columns.astype(str).str.strip()

# Nếu file chưa có tên cột theo đề và có đúng 14 cột,
# gán tên cột theo tài liệu
if "target" not in heart_df.columns and heart_df.shape[1] == 14:
    heart_df.columns = column_names

# 5. Xử lý dữ liệu thiếu
heart_df = heart_df.replace("?", np.nan)
heart_df = heart_df.dropna().copy()

# 6. Chuyển dữ liệu sang kiểu số
for col in heart_df.columns:
    heart_df[col] = pd.to_numeric(
        heart_df[col],
        errors="coerce"
    )

heart_df = heart_df.dropna().copy()

# 7. Hiển thị kết quả
print("\nKích thước dữ liệu sau tiền xử lý:",
      heart_df.shape)

print("\nTên các thuộc tính:")
print(heart_df.columns.tolist())

print("\nSố lượng dữ liệu thiếu:")
print(heart_df.isnull().sum())

print("\n5 dòng dữ liệu đầu tiên:")
display(heart_df.head())
```

```python
# BÀI 5 - CÂU 3
# KHẢO SÁT DỮ LIỆU HEART DISEASE

import pandas as pd
import matplotlib.pyplot as plt

print("Họ tên: Phạm Hải Yến")
print("MSSV: 6451071091")
print("Bài 5 - Câu 3")
print("-" * 50)

# Kích thước dữ liệu
print("Kích thước dữ liệu:", heart_df.shape)

# Thông tin dữ liệu
print("\nThông tin dữ liệu:")
heart_df.info()

# Thống kê mô tả
print("\nThống kê mô tả:")
display(heart_df.describe())

# Số lượng mẫu theo từng lớp target
print("\nSố lượng mẫu theo target:")
print(heart_df["target"].value_counts().sort_index())

# Biểu đồ phân bố target
heart_df["target"].value_counts().sort_index().plot(
    kind="bar",
    figsize=(6, 4)
)

plt.title("Phân bố target - Heart Disease")
plt.xlabel("Target")
plt.ylabel("Số lượng mẫu")
plt.xticks(rotation=0)
plt.show()
```

```python
# BÀI 5 - CÂU 4
# KHẢO SÁT TƯƠNG QUAN PEARSON

import matplotlib.pyplot as plt
import seaborn as sns

print("Họ tên: Phạm Hải Yến")
print("MSSV: 6451071091")
print("Bài 5 - Câu 4")
print("-" * 50)

# Tính ma trận tương quan
corr_matrix = heart_df.corr(method="pearson")

print("Ma trận tương quan:")
display(corr_matrix)

# Vẽ heatmap
plt.figure(figsize=(14, 10))

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Ma trận tương quan Pearson - Heart Disease")
plt.show()

# Tìm các cặp thuộc tính có tương quan cao
threshold = 0.8
high_corr = []

for i in range(len(corr_matrix.columns)):
    for j in range(i + 1, len(corr_matrix.columns)):
        value = corr_matrix.iloc[i, j]

        if abs(value) >= threshold:
            high_corr.append({
                "Thuộc tính 1": corr_matrix.columns[i],
                "Thuộc tính 2": corr_matrix.columns[j],
                "Hệ số tương quan": value
            })

print("\nCác cặp có |Pearson| >= 0.8:")

if high_corr:
    display(pd.DataFrame(high_corr))
else:
    print("Không có cặp thuộc tính nào đạt ngưỡng 0.8.")
```

```python
# BÀI 5 - CÂU 5
# TÁCH X VÀ y

print("Họ tên: Phạm Hải Yến")
print("MSSV: 6451071091")
print("Bài 5 - Câu 5")
print("-" * 50)

# X: các thuộc tính đầu vào
X = heart_df.drop(columns=["target"])

# y: thuộc tính quyết định
y = heart_df["target"]

print("Kích thước X:", X.shape)
print("Kích thước y:", y.shape)

print("\nCác thuộc tính đầu vào:")
print(X.columns.tolist())

print("\nCác lớp target:")
print(sorted(y.unique()))

print("\nSố lượng mẫu theo lớp:")
print(y.value_counts().sort_index())
```

```python
# BÀI 5 - CÂU 6
# CHIA DỮ LIỆU TRAIN VÀ TEST

from sklearn.model_selection import train_test_split

print("Họ tên: Phạm Hải Yến")
print("MSSV: 6451071091")
print("Bài 5 - Câu 6")
print("-" * 50)

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=42,
    stratify=y
)

print("Kích thước X_train:", X_train.shape)
print("Kích thước X_test:", X_test.shape)

print("Kích thước y_train:", y_train.shape)
print("Kích thước y_test:", y_test.shape)

print("\nTỷ lệ train:",
      len(X_train) / len(X))

print("Tỷ lệ test:",
      len(X_test) / len(X))
```

```python
# BÀI 5 - CÂU 7
# XÂY DỰNG CÂY QUYẾT ĐỊNH ID3

import matplotlib.pyplot as plt

from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report
)

print("Họ tên: Phạm Hải Yến")
print("MSSV: 6451071091")
print("Bài 5 - Câu 7 - ID3")
print("-" * 50)

# Khởi tạo mô hình
id3_model = DecisionTreeClassifier(
    criterion="entropy",
    random_state=42
)

# Huấn luyện
id3_model.fit(X_train, y_train)

# Dự đoán
y_pred_id3 = id3_model.predict(X_test)

# Đánh giá
accuracy_id3 = accuracy_score(
    y_test,
    y_pred_id3
)

print("Độ chính xác ID3:", accuracy_id3)

print("\nMa trận nhầm lẫn:")
print(confusion_matrix(y_test, y_pred_id3))

print("\nBáo cáo phân loại:")
print(classification_report(
    y_test,
    y_pred_id3,
    zero_division=0
))

# Vẽ cây
plt.figure(figsize=(20, 10))

plot_tree(
    id3_model,
    feature_names=X.columns,
    class_names=[str(c) for c in id3_model.classes_],
    filled=True,
    rounded=True,
    max_depth=3,
    fontsize=8
)

plt.title("Cây quyết định ID3 - Heart Disease")
plt.show()
```

```python
# BÀI 5 - CÂU 8
# XÂY DỰNG CÂY QUYẾT ĐỊNH CART

import matplotlib.pyplot as plt

from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report
)

print("Họ tên: Phạm Hải Yến")
print("MSSV: 6451071091")
print("Bài 5 - Câu 8 - CART")
print("-" * 50)

# Khởi tạo mô hình CART
cart_model = DecisionTreeClassifier(
    criterion="gini",
    random_state=42
)

# Huấn luyện
cart_model.fit(X_train, y_train)

# Dự đoán
y_pred_cart = cart_model.predict(X_test)

# Đánh giá
accuracy_cart = accuracy_score(
    y_test,
    y_pred_cart
)

print("Độ chính xác CART:", accuracy_cart)

print("\nMa trận nhầm lẫn:")
print(confusion_matrix(y_test, y_pred_cart))

print("\nBáo cáo phân loại:")
print(classification_report(
    y_test,
    y_pred_cart,
    zero_division=0
))

# Vẽ cây
plt.figure(figsize=(20, 10))

plot_tree(
    cart_model,
    feature_names=X.columns,
    class_names=[str(c) for c in cart_model.classes_],
    filled=True,
    rounded=True,
    max_depth=3,
    fontsize=8
)

plt.title("Cây quyết định CART - Heart Disease")
plt.show()
```

```python
# BÀI 5 - CÂU 9
# XÂY DỰNG MÔ HÌNH NAIVE BAYES

import matplotlib.pyplot as plt

from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report,
    ConfusionMatrixDisplay
)

print("Họ tên: Phạm Hải Yến")
print("MSSV: 6451071091")
print("Bài 5 - Câu 9 - Naive Bayes")
print("-" * 50)

# Khởi tạo mô hình
nb_model = GaussianNB()

# Huấn luyện
nb_model.fit(X_train, y_train)

# Dự đoán
y_pred_nb = nb_model.predict(X_test)

# Đánh giá
accuracy_nb = accuracy_score(
    y_test,
    y_pred_nb
)

print("Độ chính xác Naive Bayes:", accuracy_nb)

print("\nMa trận nhầm lẫn:")
cm_nb = confusion_matrix(y_test, y_pred_nb)
print(cm_nb)

print("\nBáo cáo phân loại:")
print(classification_report(
    y_test,
    y_pred_nb,
    zero_division=0
))

# Vẽ ma trận nhầm lẫn
disp = ConfusionMatrixDisplay(
    confusion_matrix=cm_nb,
    display_labels=nb_model.classes_
)

disp.plot()
plt.title("Ma trận nhầm lẫn - Naive Bayes")
plt.show()
```

```python
# BÀI 5 - CÂU 10
# SO SÁNH ID3, CART VÀ NAIVE BAYES

import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

print("Họ tên: Phạm Hải Yến")
print("MSSV: 6451071091")
print("Bài 5 - Câu 10")
print("-" * 50)

# Các mô hình và kết quả dự đoán
models = {
    "ID3": y_pred_id3,
    "CART": y_pred_cart,
    "Naive Bayes": y_pred_nb
}

# Tính các chỉ số
results = []

for model_name, y_pred in models.items():
    results.append({
        "Mô hình": model_name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(
            y_test,
            y_pred,
            average="weighted",
            zero_division=0
        ),
        "Recall": recall_score(
            y_test,
            y_pred,
            average="weighted",
            zero_division=0
        ),
        "F1-score": f1_score(
            y_test,
            y_pred,
            average="weighted",
            zero_division=0
        )
    })

comparison = pd.DataFrame(results)

print("Bảng so sánh:")
display(comparison)

# Vẽ biểu đồ Accuracy
plt.figure(figsize=(8, 5))

plt.bar(
    comparison["Mô hình"],
    comparison["Accuracy"]
)

plt.title("So sánh Accuracy - Heart Disease")
plt.xlabel("Mô hình")
plt.ylabel("Accuracy")
plt.ylim(0, 1)

for i, value in enumerate(comparison["Accuracy"]):
    plt.text(
        i,
        value + 0.01,
        f"{value:.4f}",
        ha="center"
    )

plt.show()

# Tìm mô hình có Accuracy cao nhất
best_model = comparison.loc[
    comparison["Accuracy"].idxmax()
]

print("\nMô hình có Accuracy cao nhất:",
      best_model["Mô hình"])

print("Accuracy:",
      best_model["Accuracy"])
```
